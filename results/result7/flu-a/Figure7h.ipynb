Seurat V5

```r
library(Seurat)
library(ggplot2)
```

```r
obj <- readRDS("/opt/data/private/obj_IAV.rds")
```

```r
Idents(obj) <- "celltype"
CD8 <- subset(obj,idents = "CD8+ Cytotoxic T")
```

```r
bulk.CD8 <- AggregateExpression(CD8, group.by = "disease", return.seurat = TRUE)
```

```r
Cells(bulk.CD8)
```

Heatmap换个区分度更大的颜色

```r
features <- c('PDCD1','CD244','TIGIT','LAG3',
              'IFNGR2',
              'LTB','TNF','CD40','EDAR','TNFRSF13B','TNFRSF25',
              'ACVR1C',
              'CCL4','CXCL8','CXCL10','CCR1','CCR2','CCR3','CCR6','CXCR3','CXCR4','CXCR5',
              'IL1B','IL7R','IL18BP','IL1R2','IL6R')
```

```r
options(repr.plot.width = 5, repr.plot.height = 6)
p1 <- DoHeatmap(object = bulk.CD8, features = features, angle = 30) +
   scale_fill_gradientn(colors = c("white","grey","firebrick3"))
p1
```
