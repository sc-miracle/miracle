```r
library(Seurat)
library(SeuratDisk)
library(ggplot2)
```

```r
obj <- readRDS("/opt/data/private/Update/obj_IAV.rds")
```

```r
str(obj)
```

```r
Idents(obj) <- "disease"
IAV <- subset(obj,idents = 'IAV')
```

```r
options(repr.plot.width = 10, repr.plot.height = 6)
DimPlot(IAV,label = TRUE, group.by = "celltype",raster=FALSE)
```

```r
p1 <- FeaturePlot(IAV,features = "HERC1",
  cols = c("lightgrey", "blue"),
  pt.size = 0.1,
  order = TRUE,
  min.cutoff = 0,
  max.cutoff = NA,
  slot = "count")
p1
```
